```python
def create_table(conn , table_name , df):
  conn = sqlite3.connect('pincode.db')

  dtype= pd.DataFrame({
      'column':df.columns,
      'sqlite_type':df.dtypes.values
  })

  columns = []
  for i , k in dtype.iterrows():
    column = k['column'].replace(" ","_")
    data_type = str(k['sqlite_type'])
    if data_type == "object":
        sqlite_type = "TEXT"

    elif data_type == "int64":
      sqlite_type = "INTEGER"
    columns.append(f"{column} {sqlite_type} ")
  query = f"""
  CREATE TABLE {table_name} (
        {', '.join(columns)}
  );
  """
  cursor.execute(query)
  conn.commit()
  print(query)
```

```python
create_table(conn , 'pinc' ,df)
```

```python
def ask(query):
  que = f'''
    {query}
  '''
  cursor.execute(que)
  result = cursor.fetchall()
  for row in result:
    print(row)
```
